# Ejercicio 2 – Visualización de datos con Matplotlib

En este ejercicio se realizará una inspección y visualización inicial del conjunto de datos Automobile utilizando Pandas y Matplotlib.

## 1 - Cargar el conjunto de datos

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
autos = pd.read_csv("Automobile.csv")

autos.head()
```

Comentario: El archivo Automobile.csv se cargó correctamente en un DataFrame de Pandas y se visualizaron las primeras cinco filas para comprobar su estructura.

## 2. Relación entre peso y consumo de combustible

```python
plt.scatter(autos["weight"], autos["mpg"])

plt.xlabel("Peso del automóvil (weight)")
plt.ylabel("Millas por galón (mpg)")
plt.title("Relación entre peso y consumo de combustible")

plt.show()
```

Comentario: Se observa una relación negativa entre el peso del automóvil y las millas por galón (mpg). A medida que aumenta el peso, disminuye el rendimiento de combustible.

## 3. Distribución del consumo de combustible (mpg)

```python
plt.hist(autos["mpg"], bins=10, edgecolor="black")

plt.xlabel("Millas por galón (mpg)")
plt.ylabel("Frecuencia")
plt.title("Distribución del consumo de combustible")

plt.show()
```

Comentario: La mayor parte de los automóviles presenta valores de consumo entre aproximadamente 15 y 30 mpg. Los valores más altos de mpg son menos frecuentes.

## 4. Diagramas de caja de cylinders y origin

```python
autos["origin"].unique()
```

```python
origin_numerico = autos["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})

plt.boxplot([autos["cylinders"], origin_numerico])

plt.xticks([1, 2], ["Cylinders", "Origin"])
plt.ylabel("Valor")
plt.title("Diagramas de caja de cylinders y origin")

plt.show()
```

**Comentario:** El diagrama de caja de `cylinders` muestra que los automóviles presentan principalmente entre 4 y 8 cilindros. Para `origin`, los valores 1, 2 y 3 representan categorías de origen y no una variable numérica continua, por lo que el boxplot debe interpretarse con precaución.

## 5. Distribución de automóviles según año del modelo y origen

```python
conteo = autos.groupby(["model_year", "origin"]).size().unstack(fill_value=0)

conteo.plot(kind="bar")

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución de automóviles según año del modelo y origen")
plt.legend(title="Origen")

plt.show()
```

**Comentario:** Se observa un predominio de automóviles de origen estadounidense en la mayoría de los años analizados. Los automóviles europeos y japoneses presentan cantidades menores, aunque hacia los últimos años aumenta la presencia relativa de vehículos japoneses.
