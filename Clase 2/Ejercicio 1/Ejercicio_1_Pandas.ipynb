# Ejercicio 1 – Carga y análisis de archivos con Pandas

En este ejercicio se trabajará con archivos CSV, JSON y Excel utilizando la biblioteca Pandas.

```python
import pandas as pd
```

## 1. Carga de datos de ventas desde CSV

```python
ventas = pd.read_csv("ventas.csv")

ventas.head()
```

**Comentario:** El archivo `ventas.csv` se cargó correctamente mediante Pandas y se mostraron las primeras cinco filas como vista previa.

## 2. Carga de datos de clientes desde JSON

```python
clientes = pd.read_json("clientes.json")

clientes.head()
```

```python
cantidad_clientes = len(clientes)

print("Cantidad total de clientes:", cantidad_clientes)
```

**Comentario:** El archivo `clientes.json` se cargó correctamente mediante Pandas. La cantidad total de clientes es 6.

## 3. Carga de datos de inventario desde Excel

```python
inventario = pd.read_excel("inventario.xlsx")

inventario.head()
```

```python
promedio_stock = inventario["stock_disponible"].mean()

print("Promedio de stock disponible:", promedio_stock)
```

**Comentario:** El archivo `inventario.xlsx` se cargó correctamente mediante Pandas. El promedio de stock disponible de todos los productos es de 25 unidades.
